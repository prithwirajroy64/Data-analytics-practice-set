### Your task:
1. Load the data using Pandas.
2. Inspect the first five rows.
3. Find the number of rows and columns.
4. Filter only paid orders.
5. Calculate revenue for each order.
6. Group the orders by category.
7. Calculate the total revenue of each category.
8. Export the result into a CSV file.

```python
import pandas as pd
from io import StringIO
```

```python
csv_data = """
order_id,category,status,quantity,unit_price
101,Electronics,paid,2,499.50
102,Home,pending,1,200.00
103,Electronics,paid,1,999.00
104,Home,paid,3,100.00
105,Books,paid,2,250.00
"""
```

```python
# 1 & 2 task are done
df = pd.read_csv(StringIO(csv_data))
# df = pd.read_csv("orders.csv")  --> in real world senario
print(df.head())
```

```python
# 3. number of rows and columns
print(df.shape) #shape: number of rows and columns.
```

```python
print(df.info()) #info(): column data types and missing values.
```

```python
print(df.describe()) #describe(): statistics for numeric columns.
```

```python
# 4. Filter paid orders
paid_orders = df.loc[
  df["status"] == "paid"
].copy()

paid_orders
# Why .copy()?
# It creates an independent DataFrame for subsequent transformations and avoids common chained-assignment problems.
```

```python
# 5. revenue for each order
paid_orders["revenue"] = (
  paid_orders["quantity"] * paid_orders["unit_price"]
)
paid_orders["revenue"]
```

```python
# 6. Group the orders by category
summary = (
  paid_orders.groupby("category", as_index=False).agg(
    paid_orders = ("order_id", "count"),
    total_revenue = ("revenue", "sum")
  ).sort_values(
    "total_revenue",
    ascending=False
  )
)

print(summary)
```

```python
#7. Calculate the total revenue of each category.

total_revenue = paid_orders["revenue"].sum()
total_paid_orders = len(paid_orders)

print(f"Total paid orders: {total_paid_orders}")
print(f"Total revenue:{total_revenue}")
```

```python
# 8. Export the result into a CSV file.
```

summary.to_csv(
    "category_sales_summary.csv",
    index=False
)
