# Data Transformation & Validation

```python
orders = [
    {
        "order_id": "A101",
        "status": "paid",
        "quantity": "2",
        "unit_price": "499.50"
    },
    {
        "order_id": "A102",
        "status": "pending",
        "quantity": "1",
        "unit_price": "200.00"
    },
    {
        "status": "paid",
        "quantity": "3",
    },
    {
        "order_id": "A104",
        "status": "paid",
        "quantity": None,
        "unit_price": "50.00"
    },
    {
        "order_id": "A105",
        "status": "paid",
        "quantity": "-1",
        "unit_price": "80.00"
    }
]
```

---

# My practical assignment

- Accept the list of orders.
- Process only orders with status paid.
- Convert quantity into an integer.
- Convert price into a decimal number.
- Reject records with missing values, invalid numbers or negative quantity.
- Calculate total revenue.
- Return the number of successfully processed orders, revenue and rejected order IDs.

Expected output:

```python
{
    "paid_orders_processed": 2,
    "revenue": "1299.00",
    "rejected_order_ids": ["A104", "A105"]
}
```

---

```python
from decimal import InvalidOperation, Decimal
```

```python
def summarize_paid_orders(orders:list):

  paid_orders_processed=0
  revenue=Decimal("0.00")
  rejected_order_ids=[]


  for order in orders:

    # ignore those orders which are not paid
    if order.get("status") != "paid":
          continue


    try:
      quantity = int(order["quantity"])
      price = Decimal(order["unit_price"])

      # maintain order logic or rules
      if quantity < 0 or price < 0:
        raise ValueError("Invalid order values.")
      
      revenue +=  quantity * price #this is the revenue business logic
      paid_orders_processed += 1
    
    except (
      KeyError,
      ValueError,
      TypeError,
      InvalidOperation
    ):
      rejected_order_ids.append(order.get("order_id", "UNKNOWN")) #if any order not have order id then that is mark as UNKNOWN

  return {
              "paid_orders_processed": paid_orders_processed,
              "revenue": f"{revenue:.2f}",
              "rejected_order_ids": rejected_order_ids
          }

    
```

```python
result = summarize_paid_orders(orders)
print(result)
```
